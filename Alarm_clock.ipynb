```python
import datetime  # Importing datetime to get the current system time
import time      # Importing time to add delay in the loop
import playsound # Importing playsound to play the alarm sound

# Function to validate the user input time
# - Ensures the hour is between 1 and 12
# - Ensures the minute is between 0 and 59
# - Ensures the period is either 'am' or 'pm'

def validate_time(hour, minute, period):
    if not (1 <= hour <= 12):
        print("Invalid hour! Enter a value between 1 and 12.")
        return False
    if not (0 <= minute <= 59):
        print("Invalid minute! Enter a value between 0 and 59.")
        return False
    if period not in ["am", "pm"]:
        print("Invalid period! Enter 'am' or 'pm'.")
        return False
    return True

# Function to get the alarm time from the user
# - Keeps asking until valid input is received

def set_alarm():
    while True:
        try:
            # Asking user to enter hour, minute, and period (AM/PM)
            hour = int(input("Enter Hour (1-12): "))
            minute = int(input("Enter Minutes (0-59): "))
            period = input("Enter am / pm: ").strip().lower()

            # Validate input time
            if validate_time(hour, minute, period):
                return hour, minute, period
        except ValueError:
            print("Invalid input! Please enter numbers for hour and minute.")

# Function to play the alarm sound
# - Handles errors if the sound file is missing or cannot be played

def play_alarm():
    try:
        playsound.playsound("alarm.mp3")
    except Exception as e:
        print(f"Error playing sound: {e}")

# Main function to run the alarm clock
# - Converts 12-hour time to 24-hour format
# - Waits for the alarm time to match the system time
# - Plays the alarm sound when time is reached
# - Handles keyboard interruption gracefully

def main():
    # Get alarm time from the user
    hour, minute, period = set_alarm()

     # Convert 12-hour format to 24-hour format
    if period == "pm" and hour != 12:
        hour += 12 # Convert PM hours (except 12 PM)
    elif period == "am" and hour == 12:
        hour = 0   # Convert 12 AM to 00 (midnight)

    # Display the set alarm time in 24-hour format
    print(f"Alarm set for {hour:02d}:{minute:02d}")

    try:
        while True:
            # Get the current system time
            now = datetime.datetime.now()


            # Check if the current time matches the alarm time
            if now.hour == hour and now.minute == minute:
                print("Playing alarm...")
                play_alarm() # Play the alarm sound
                break        # Exit the loop after playing the alarm
            time.sleep(1)    # Wait 1 second before checking again (reduces CPU usage)
    except KeyboardInterrupt:
        # Gracefully exit if the user stops the program manually
        print("\nAlarm canceled by user. Exiting...")

# Run the program when executed
if __name__ == "__main__":
    main()
```
